# E-commerce Sales & Profitability Analysis

**Goal:** Analyze orders, products, customers, and time-based performance using Pandas.

### Business questions
1. What are the overall revenue, cost, profit, and profit margin?
2. Which products drive revenue and profit?
3. How do repeat and one-time customers differ?
4. How does performance change by month and year?
5. Which products and periods show the strongest growth?
6. Which customers contribute most to revenue and profit?

> **Data source:** `orders_full.csv`  
> Keep the CSV in the same folder as this notebook, or update `DATA_PATH` below.


```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

DATA_PATH = Path("orders_full.csv")

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"{DATA_PATH} was not found. Place the CSV in the notebook folder "
        "or update DATA_PATH."
    )

df = pd.read_csv(DATA_PATH)
df.head()
```

## 1. Data inspection

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
display(df.isna().sum().sort_values(ascending=False))

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df.info()
```

```python
df.describe(include="all").T
```

## 2. Data cleaning and validation

```python
# Convert timestamp
df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce")

# Validate required fields
required_columns = [
    "order_id", "user_id", "primary_product_id",
    "created_at", "items_purchased", "price_usd", "cogs_usd"
]

missing_required = [col for col in required_columns if col not in df.columns]
if missing_required:
    raise ValueError(f"Missing required columns: {missing_required}")

# Check invalid numeric values
numeric_cols = ["items_purchased", "price_usd", "cogs_usd"]
display(df[numeric_cols].isna().sum().rename("missing_values").to_frame())

print("Invalid dates:", df["created_at"].isna().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate order IDs:", df["order_id"].duplicated().sum())
```

```python
# Keep rows required for the core profitability analysis.
df = df.dropna(subset=required_columns).copy()

# Quantity, price and cost should not be negative.
invalid_values = (
    (df["items_purchased"] <= 0)
    | (df["price_usd"] < 0)
    | (df["cogs_usd"] < 0)
)

print("Rows with invalid quantity/price/cost:", invalid_values.sum())

df = df.loc[~invalid_values].copy()
print("Shape after validation:", df.shape)
```

## 3. Feature engineering

```python
df["revenue"] = df["items_purchased"] * df["price_usd"]
df["cost"] = df["items_purchased"] * df["cogs_usd"]
df["profit"] = df["revenue"] - df["cost"]

df["profit_margin_pct"] = np.where(
    df["revenue"] != 0,
    df["profit"] / df["revenue"] * 100,
    np.nan
)

df["date"] = df["created_at"].dt.date
df["year"] = df["created_at"].dt.year
df["month"] = df["created_at"].dt.month
df["month_name"] = df["created_at"].dt.month_name()
df["day_name"] = df["created_at"].dt.day_name()

df[[
    "order_id", "created_at", "items_purchased",
    "price_usd", "cogs_usd", "revenue", "cost",
    "profit", "profit_margin_pct"
]].head()
```

```python
# Core data-quality checks after feature engineering
quality_checks = pd.Series({
    "rows": len(df),
    "unique_orders": df["order_id"].nunique(),
    "unique_customers": df["user_id"].nunique(),
    "unique_products": df["primary_product_id"].nunique(),
    "missing_revenue": df["revenue"].isna().sum(),
    "missing_profit": df["profit"].isna().sum(),
    "negative_revenue": (df["revenue"] < 0).sum(),
})

quality_checks
```

## 4. Overall business performance

```python
total_revenue = df["revenue"].sum()
total_cost = df["cost"].sum()
total_profit = df["profit"].sum()
total_orders = df["order_id"].nunique()
total_customers = df["user_id"].nunique()
total_units = df["items_purchased"].sum()

overall = pd.Series({
    "Revenue": total_revenue,
    "Cost": total_cost,
    "Profit": total_profit,
    "Profit Margin (%)": total_profit / total_revenue * 100,
    "Orders": total_orders,
    "Customers": total_customers,
    "Units Sold": total_units,
    "Average Order Value": total_revenue / total_orders,
    "Profit per Order": total_profit / total_orders,
})

overall.round(2)
```

## 5. Product performance

```python
product_summary = (
    df.groupby("primary_product_id")
      .agg(
          orders=("order_id", "nunique"),
          items_sold=("items_purchased", "sum"),
          revenue=("revenue", "sum"),
          cost=("cost", "sum"),
          profit=("profit", "sum"),
      )
      .reset_index()
)

product_summary["profit_margin_pct"] = (
    product_summary["profit"] / product_summary["revenue"] * 100
)

product_summary["revenue_share_pct"] = (
    product_summary["revenue"] / product_summary["revenue"].sum() * 100
)

product_summary["profit_share_pct"] = (
    product_summary["profit"] / product_summary["profit"].sum() * 100
)

product_summary["items_per_order"] = (
    product_summary["items_sold"] / product_summary["orders"]
)

product_summary["revenue_per_order"] = (
    product_summary["revenue"] / product_summary["orders"]
)

product_summary.sort_values("revenue", ascending=False).round(2)
```

```python
print("Top products by revenue")
display(
    product_summary[
        ["primary_product_id", "revenue", "revenue_share_pct"]
    ].sort_values("revenue", ascending=False).head(10).round(2)
)

print("Top products by profit")
display(
    product_summary[
        ["primary_product_id", "profit", "profit_share_pct"]
    ].sort_values("profit", ascending=False).head(10).round(2)
)

print("Top products by profit margin")
display(
    product_summary[
        ["primary_product_id", "profit_margin_pct"]
    ].sort_values("profit_margin_pct", ascending=False).head(10).round(2)
)
```

```python
avg_product_revenue = product_summary["revenue"].mean()
avg_product_margin = product_summary["profit_margin_pct"].mean()

product_summary["performance"] = np.select(
    [
        (product_summary["revenue"] >= avg_product_revenue)
        & (product_summary["profit_margin_pct"] >= avg_product_margin),

        (product_summary["revenue"] >= avg_product_revenue)
        & (product_summary["profit_margin_pct"] < avg_product_margin),

        (product_summary["revenue"] < avg_product_revenue)
        & (product_summary["profit_margin_pct"] >= avg_product_margin),
    ],
    [
        "High Revenue - High Margin",
        "High Revenue - Low Margin",
        "Low Revenue - High Margin",
    ],
    default="Low Revenue - Low Margin",
)

product_summary[
    ["primary_product_id", "revenue", "profit", "profit_margin_pct", "performance"]
].sort_values("profit", ascending=False).round(2)
```

## 6. Customer analysis

```python
customer_summary = (
    df.groupby("user_id")
      .agg(
          orders=("order_id", "nunique"),
          units_purchased=("items_purchased", "sum"),
          revenue=("revenue", "sum"),
          cost=("cost", "sum"),
          profit=("profit", "sum"),
      )
      .reset_index()
)

customer_summary["profit_margin_pct"] = (
    customer_summary["profit"] / customer_summary["revenue"] * 100
)

customer_summary["avg_order_value"] = (
    customer_summary["revenue"] / customer_summary["orders"]
)

customer_summary["customer_type"] = np.where(
    customer_summary["orders"] > 1,
    "Repeat",
    "One-time"
)

customer_summary.head()
```

```python
customer_type_summary = (
    customer_summary.groupby("customer_type")
    .agg(
        customers=("user_id", "nunique"),
        orders=("orders", "sum"),
        revenue=("revenue", "sum"),
        profit=("profit", "sum"),
    )
    .reset_index()
)

customer_type_summary["customer_share_pct"] = (
    customer_type_summary["customers"]
    / customer_type_summary["customers"].sum() * 100
)

customer_type_summary["revenue_share_pct"] = (
    customer_type_summary["revenue"]
    / customer_type_summary["revenue"].sum() * 100
)

customer_type_summary["profit_share_pct"] = (
    customer_type_summary["profit"]
    / customer_type_summary["profit"].sum() * 100
)

customer_type_summary["revenue_per_customer"] = (
    customer_type_summary["revenue"]
    / customer_type_summary["customers"]
)

customer_type_summary["profit_per_customer"] = (
    customer_type_summary["profit"]
    / customer_type_summary["customers"]
)

customer_type_summary.round(2)
```

```python
repeat_customer_rate = (
    (customer_summary["customer_type"] == "Repeat").mean() * 100
)

print(f"Repeat customer rate: {repeat_customer_rate:.2f}%")

print("\nTop customers by revenue")
display(
    customer_summary
    .sort_values("revenue", ascending=False)
    .head(10)
    .round(2)
)

print("Top customers by profit")
display(
    customer_summary
    .sort_values("profit", ascending=False)
    .head(10)
    .round(2)
)
```

### Customer revenue concentration (Pareto analysis)

```python
customer_pareto = (
    customer_summary
    .sort_values("revenue", ascending=False)
    .reset_index(drop=True)
    .copy()
)

customer_pareto["cumulative_revenue"] = customer_pareto["revenue"].cumsum()
customer_pareto["cumulative_revenue_pct"] = (
    customer_pareto["cumulative_revenue"]
    / customer_pareto["revenue"].sum() * 100
)

for target in [50, 80, 90]:
    customers_needed = (
        customer_pareto["cumulative_revenue_pct"].ge(target).idxmax() + 1
    )
    customer_pct = customers_needed / len(customer_pareto) * 100
    print(
        f"{target}% of revenue requires {customers_needed:,} customers "
        f"({customer_pct:.2f}% of customers)"
    )
```

## 7. Monthly and yearly performance

```python
monthly_summary = (
    df.groupby(["year", "month", "month_name"])
      .agg(
          orders=("order_id", "nunique"),
          customers=("user_id", "nunique"),
          units_sold=("items_purchased", "sum"),
          revenue=("revenue", "sum"),
          profit=("profit", "sum"),
      )
      .reset_index()
      .sort_values(["year", "month"])
)

monthly_summary["profit_margin_pct"] = (
    monthly_summary["profit"] / monthly_summary["revenue"] * 100
)

monthly_summary["avg_order_value"] = (
    monthly_summary["revenue"] / monthly_summary["orders"]
)

monthly_summary["period"] = pd.to_datetime(
    monthly_summary["year"].astype(str)
    + "-"
    + monthly_summary["month"].astype(str).str.zfill(2)
    + "-01"
)

monthly_summary.round(2)
```

```python
best_revenue_month = monthly_summary.loc[monthly_summary["revenue"].idxmax()]
worst_revenue_month = monthly_summary.loc[monthly_summary["revenue"].idxmin()]
best_profit_month = monthly_summary.loc[monthly_summary["profit"].idxmax()]
worst_profit_month = monthly_summary.loc[monthly_summary["profit"].idxmin()]
highest_margin_month = monthly_summary.loc[monthly_summary["profit_margin_pct"].idxmax()]
lowest_margin_month = monthly_summary.loc[monthly_summary["profit_margin_pct"].idxmin()]

summary_months = pd.DataFrame({
    "metric": [
        "Highest Revenue",
        "Lowest Revenue",
        "Highest Profit",
        "Lowest Profit",
        "Highest Profit Margin",
        "Lowest Profit Margin",
    ],
    "year": [
        best_revenue_month["year"],
        worst_revenue_month["year"],
        best_profit_month["year"],
        worst_profit_month["year"],
        highest_margin_month["year"],
        lowest_margin_month["year"],
    ],
    "month": [
        best_revenue_month["month_name"],
        worst_revenue_month["month_name"],
        best_profit_month["month_name"],
        worst_profit_month["month_name"],
        highest_margin_month["month_name"],
        lowest_margin_month["month_name"],
    ],
    "value": [
        best_revenue_month["revenue"],
        worst_revenue_month["revenue"],
        best_profit_month["profit"],
        worst_profit_month["profit"],
        highest_margin_month["profit_margin_pct"],
        lowest_margin_month["profit_margin_pct"],
    ],
})

summary_months.round(2)
```

```python
yearly_summary = (
    monthly_summary.groupby("year")
    .agg(
        orders=("orders", "sum"),
        units_sold=("units_sold", "sum"),
        revenue=("revenue", "sum"),
        profit=("profit", "sum"),
    )
    .reset_index()
)

yearly_summary["profit_margin_pct"] = (
    yearly_summary["profit"] / yearly_summary["revenue"] * 100
)

yearly_summary["avg_order_value"] = (
    yearly_summary["revenue"] / yearly_summary["orders"]
)

yearly_summary["profit_per_order"] = (
    yearly_summary["profit"] / yearly_summary["orders"]
)

yearly_summary["aov_growth_pct"] = (
    yearly_summary["avg_order_value"].pct_change() * 100
)

yearly_summary.round(2)
```

## 8. Year-over-year monthly growth

```python
monthly_yoy = monthly_summary.copy()

monthly_yoy["revenue_yoy_pct"] = (
    monthly_yoy.groupby("month")["revenue"].pct_change() * 100
)

monthly_yoy["profit_yoy_pct"] = (
    monthly_yoy.groupby("month")["profit"].pct_change() * 100
)

monthly_yoy[
    [
        "year", "month_name", "revenue", "profit",
        "revenue_yoy_pct", "profit_yoy_pct"
    ]
].round(2)
```

```python
print("Highest YoY revenue growth:")
display(
    monthly_yoy.loc[
        monthly_yoy["revenue_yoy_pct"].idxmax(),
        ["year", "month_name", "revenue_yoy_pct"]
    ].to_frame().T.round(2)
)

print("Lowest YoY revenue growth:")
display(
    monthly_yoy.loc[
        monthly_yoy["revenue_yoy_pct"].idxmin(),
        ["year", "month_name", "revenue_yoy_pct"]
    ].to_frame().T.round(2)
)
```

## 9. Product growth by year

```python
product_yearly = (
    df.groupby(["year", "primary_product_id"])
      .agg(
          orders=("order_id", "nunique"),
          items_sold=("items_purchased", "sum"),
          revenue=("revenue", "sum"),
          profit=("profit", "sum"),
      )
      .reset_index()
)

product_yearly["profit_margin_pct"] = (
    product_yearly["profit"] / product_yearly["revenue"] * 100
)

product_yearly["revenue_growth_pct"] = (
    product_yearly.groupby("primary_product_id")["revenue"].pct_change() * 100
)

product_yearly["profit_growth_pct"] = (
    product_yearly.groupby("primary_product_id")["profit"].pct_change() * 100
)

product_yearly.round(2)
```

```python
# If the dataset contains 2013 and 2014, compare those years directly.
available_growth_years = [y for y in [2013, 2014] if y in product_yearly["year"].unique()]

if len(available_growth_years) == 2:
    growth_data = product_yearly[
        product_yearly["year"].isin(available_growth_years)
    ].copy()

    revenue_pivot = growth_data.pivot(
        index="primary_product_id",
        columns="year",
        values="revenue"
    )

    profit_pivot = growth_data.pivot(
        index="primary_product_id",
        columns="year",
        values="profit"
    )

    old_year, new_year = available_growth_years

    revenue_pivot["revenue_increase"] = (
        revenue_pivot[new_year] - revenue_pivot[old_year]
    )
    revenue_pivot["revenue_growth_pct"] = (
        revenue_pivot["revenue_increase"] / revenue_pivot[old_year] * 100
    )

    profit_pivot["profit_increase"] = (
        profit_pivot[new_year] - profit_pivot[old_year]
    )
    profit_pivot["profit_growth_pct"] = (
        profit_pivot["profit_increase"] / profit_pivot[old_year] * 100
    )

    print("Revenue growth:")
    display(revenue_pivot.sort_values("revenue_increase", ascending=False).round(2))

    print("Profit growth:")
    display(profit_pivot.sort_values("profit_increase", ascending=False).round(2))
else:
    print("Direct 2013 vs 2014 comparison skipped because both years are not present.")
```

## 12. Final business insights checklist

After running the notebook, write the conclusions from the actual output rather than hard-coding them:

- **Overall:** revenue, profit, margin, orders, customers, and AOV.
- **Products:** products with the highest revenue, highest profit, and highest margin.
- **Customers:** repeat-customer rate and revenue/profit contribution of repeat vs one-time customers.
- **Concentration:** percentage of customers needed to generate 80% of revenue.
- **Time:** highest/lowest revenue and profit months.
- **Growth:** strongest YoY revenue/profit changes.
- **Product growth:** products with the largest revenue/profit increase between comparable years.

### Important analytical notes
- `revenue = quantity × selling price`
- `cost = quantity × COGS per unit`
- `profit = revenue − cost`
- `profit margin = profit / revenue × 100`
- Revenue/profit growth should be interpreted only when the comparison periods are comparable.
- Product IDs are identifiers; without a product-name lookup table, do not invent product names.
